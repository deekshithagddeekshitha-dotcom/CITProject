```python

import os
import re
import json
import requests
import fitz
import gradio as gr
import numpy as np
import faiss

from bs4 import BeautifulSoup
from datetime import datetime
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# =====================================
# 1. LOAD LOCAL AI MODEL
# =====================================

print("Loading TruthLens AI model...")

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

HEADERS = {
    "User-Agent": "Mozilla/5.0"
}

# =====================================
# 2. WEBSITE EXTRACTION
# =====================================

def extract_website(url):

    try:

        if not url.startswith(("http://", "https://")):
            url = "https://" + url

        response = requests.get(
            url,
            headers=HEADERS,
            timeout=15
        )

        response.raise_for_status()

        soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        for tag in soup([
            "script", "style", "nav",
            "footer", "header", "noscript"
        ]):
            tag.decompose()

        title = (
            soup.title.get_text(strip=True)
            if soup.title else url
        )

        text = soup.get_text(
            separator=" ",
            strip=True
        )

        return {
            "name": title,
            "url": url,
            "text": text[:40000],
            "type": "Website"
        }

    except Exception as e:

        return {
            "name": url,
            "url": url,
            "text": "",
            "type": "Website",
            "error": str(e)
        }


# =====================================
# 3. PDF EXTRACTION
# =====================================

def extract_pdf(filepath):

    try:

        document = fitz.open(filepath)

        text = "\n".join(
            page.get_text()
            for page in document
        )

        return {
            "name": os.path.basename(filepath),
            "url": "Uploaded PDF",
            "text": text[:40000],
            "type": "PDF"
        }

    except Exception as e:

        return {
            "name": os.path.basename(filepath),
            "url": "Uploaded PDF",
            "text": "",
            "type": "PDF",
            "error": str(e)
        }


# =====================================
# 4. CHUNK CREATION
# =====================================

def create_chunks(sources):

    chunks = []

    for source in sources:

        sentences = re.split(
            r'(?<=[.!?])\s+',
            source["text"]
        )

        current = ""

        for sentence in sentences:

            if len(current) + len(sentence) < 1000:
                current += sentence + " "

            else:

                if len(current.strip()) > 30:

                    chunks.append({
                        "source": source["name"],
                        "url": source["url"],
                        "text": current.strip()
                    })

                current = sentence + " "

        if len(current.strip()) > 30:

            chunks.append({
                "source": source["name"],
                "url": source["url"],
                "text": current.strip()
            })

    return chunks


# =====================================
# 5. FAISS VECTOR SEARCH
# =====================================

def create_vector_database(chunks):

    if not chunks:
        return None, None

    texts = [
        item["text"] for item in chunks
    ]

    embeddings = model.encode(
        texts,
        convert_to_numpy=True
    ).astype("float32")

    faiss.normalize_L2(embeddings)

    index = faiss.IndexFlatIP(
        embeddings.shape[1]
    )

    index.add(embeddings)

    return index, embeddings


def semantic_search(question, chunks, index):

    if index is None:
        return []

    query_vector = model.encode(
        [question],
        convert_to_numpy=True
    ).astype("float32")

    faiss.normalize_L2(query_vector)

    scores, indices = index.search(
        query_vector,
        min(10, len(chunks))
    )

    results = []

    for score, idx in zip(scores[0], indices[0]):

        if idx >= 0:

            item = chunks[idx].copy()

            item["score"] = round(
                float(score), 3
            )

            results.append(item)

    return results


# =====================================
# 6. TOPIC KEYWORD GENERATOR
# =====================================

def get_expected_topics(topic):

    topic = topic.lower()

    if "scholarship" in topic:

        return [
            "eligibility criteria",
            "application deadline",
            "required documents",
            "renewal conditions",
            "application procedure",
            "selection criteria",
            "financial benefits"
        ]

    elif "college" in topic:

        return [
            "admission requirements",
            "college fees",
            "accreditation",
            "placement information",
            "faculty details",
            "hostel facilities",
            "course details"
        ]

    elif "government scheme" in topic:

        return [
            "eligibility",
            "scheme benefits",
            "required documents",
            "application procedure",
            "deadline",
            "renewal conditions"
        ]

    elif "business" in topic:

        return [
            "investment",
            "market demand",
            "competition",
            "customer requirements",
            "business risks",
            "legal requirements"
        ]

    return [
        "requirements",
        "cost details",
        "important dates",
        "risks and limitations",
        "implementation process",
        "supporting evidence"
    ]


# =====================================
# 7. INFORMATION GAP DETECTOR
# =====================================

def detect_gaps(topic, chunks):

    expected = get_expected_topics(topic)

    texts = [
        item["text"] for item in chunks
    ]

    if not texts:
        return expected

    embeddings = model.encode(
        expected + texts,
        convert_to_numpy=True
    )

    expected_embeddings = embeddings[:len(expected)]
    source_embeddings = embeddings[len(expected):]

    similarities = cosine_similarity(
        expected_embeddings,
        source_embeddings
    )

    gaps = []

    for i, item in enumerate(expected):

        best_score = float(
            similarities[i].max()
        )

        if best_score < 0.35:

            gaps.append({
                "topic": item,
                "similarity": round(best_score, 3),
                "status": "Potential information gap"
            })

    return gaps


# =====================================
# 8. CONFLICT DETECTOR
# =====================================

def detect_conflicts(chunks):

    conflicts = []

    if len(chunks) < 2:
        return conflicts

    date_pattern = re.compile(
        r'\b(?:\d{1,2}[/-]\d{1,2}[/-]\d{2,4}|'
        r'\d{1,2}\s+'
        r'(?:January|February|March|April|May|June|'
        r'July|August|September|October|November|December)'
        r'\s+20\d{2})\b',
        re.IGNORECASE
    )

    grouped = {}

    for item in chunks:

        for date in date_pattern.findall(item["text"]):

            key = (
                item["source"],
                item["url"]
            )

            grouped.setdefault(key, set()).add(date)

    items = list(grouped.items())

    for i in range(len(items)):

        for j in range(i + 1, len(items)):

            source_a, dates_a = items[i]
            source_b, dates_b = items[j]

            if source_a[1] != source_b[1]:

                if dates_a != dates_b:

                    conflicts.append({
                        "source_1": source_a[0],
                        "dates_1": list(dates_a),
                        "source_2": source_b[0],
                        "dates_2": list(dates_b),
                        "status": "Possible date discrepancy - verify context"
                    })

    return conflicts


# =====================================
# 9. FRESHNESS CHECKER
# =====================================

def check_freshness(sources):

    results = []

    current_year = datetime.now().year

    for source in sources:

        years = re.findall(
            r'\b20\d{2}\b',
            source["text"]
        )

        if years:

            latest = max(map(int, years))

            if latest < current_year - 1:
                status = "Potentially outdated"

            elif latest < current_year:
                status = "Needs verification"

            else:
                status = "Recent year mentioned"

        else:

            latest = None
            status = "Date not identified"

        results.append({
            "source": source["name"],
            "year": latest,
            "status": status
        })

    return results


# =====================================
# 10. VERIFICATION CHECKLIST
# =====================================

def create_checklist(gaps, conflicts, freshness):

    checklist = []

    for item in gaps:

        checklist.append({
            "priority": "High",
            "task": "Verify " + item["topic"]
        })

    for item in conflicts:

        checklist.append({
            "priority": "High",
            "task": (
                "Verify the date discrepancy between "
                + item["source_1"]
                + " and "
                + item["source_2"]
            )
        })

    for item in freshness:

        if (
            "outdated" in item["status"].lower()
            or "not identified" in item["status"].lower()
        ):

            checklist.append({
                "priority": "Medium",
                "task": "Verify publication date of " + item["source"]
            })

    if not checklist:

        checklist.append({
            "priority": "Medium",
            "task": "Review official sources for completeness"
        })

    return checklist


# =====================================
# 11. MAIN INVESTIGATION
# =====================================

def investigate(topic, urls, pdf_files):

    if not topic.strip():

        return "Please enter a research question."

    sources = []

    url_list = [
        x.strip()
        for x in (urls or "").splitlines()
        if x.strip()
    ]

    for url in url_list:

        sources.append(
            extract_website(url)
        )

    if pdf_files:

        if isinstance(pdf_files, str):
            pdf_files = [pdf_files]

        for file in pdf_files:

            sources.append(
                extract_pdf(file)
            )

    sources = [
        s for s in sources
        if s["text"].strip()
    ]

    if not sources:

        return "No readable sources found."

    chunks = create_chunks(sources)

    if not chunks:

        return "Not enough text extracted."

    index, embeddings = create_vector_database(chunks)

    evidence = semantic_search(
        topic,
        chunks,
        index
    )

    gaps = detect_gaps(
        topic,
        chunks
    )

    conflicts = detect_conflicts(chunks)

    freshness = check_freshness(sources)

    checklist = create_checklist(
        gaps,
        conflicts,
        freshness
    )

    # =================================
    # REPORT
    # =================================

    report = f"""
# 🔍 TRUTHLENS AI

### AI-Powered Information Investigation Platform

**Research Question:** {topic}

**Date:** {datetime.now().strftime('%d-%m-%Y')}

---

## 1. Investigation Overview

- Sources reviewed: {len(sources)}
- Evidence chunks: {len(chunks)}
- Semantic search: Enabled
- Vector database: FAISS
- External AI API: Not required

---

## 2. Source Collection

"""

    for s in sources:

        report += f"""
**{s['name']}**

- Type: {s['type']}
- Reference: {s['url']}

"""

    report += """
---

## 3. Information Gap Detection

"""

    if gaps:

        for gap in gaps:

            report += (
                f"- ⚠️ **{gap['topic']}** "
                f"(Similarity: {gap['similarity']})\n"
            )

    else:

        report += "No potential gaps identified by the detector.\n"

    report += """
---

## 4. Possible Conflicts

"""

    if conflicts:

        for item in conflicts:

            report += f"""
- **{item['source_1']}**: {item['dates_1']}
- **{item['source_2']}**: {item['dates_2']}
- {item['status']}

"""

    else:

        report += (
            "No date discrepancies identified by the basic "
            "conflict detector. This does not prove that "
            "the sources contain no other contradictions.\n"
        )

    report += """
---

## 5. Source Freshness

"""

    for item in freshness:

        report += f"""
- **{item['source']}**
- Year found: {item['year']}
- Status: {item['status']}

"""

    report += """
---

## 6. Relevant Evidence

"""

    for item in evidence:

        report += f"""
### {item['source']}

Similarity: {item['score']}

Reference: {item['url']}

> {item['text']}

"""

    report += """
---

## 7. Verification Checklist

"""

    for i, item in enumerate(checklist, 1):

        report += (
            f"{i}. **{item['priority']}** - "
            f"{item['task']}\n"
        )

    report += """

---

## 8. Investigation Limitations

- Findings depend on the collected source content.
- Similarity scores are not factual confidence scores.
- Date discrepancies are not necessarily contradictions.
- The gap detector uses predefined topic categories.
- No automated claim-level factual verification is performed.

Always verify important information through official sources.

**Generated by TruthLens AI**
"""

    return report


# =====================================
# 12. GRADIO INTERFACE
# =====================================

CSS = """
.gradio-container {
    max-width: 1100px !important;
    margin: auto !important;
}

.title {
    text-align: center;
    color: #8065ff;
}
"""

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="violet",
        neutral_hue="slate"
    ),
    css=CSS,
    title="TruthLens AI"
) as demo:

    gr.Markdown("""
    <div class="title">

    # 🔍 TRUTHLENS AI

    ### Don't just find information. Discover what's missing.

    AI-Powered Information Investigation Platform

    </div>
    """)

    gr.Markdown("""
    Enter your research question and provide multiple sources.
    No API key required.
    """)

    topic = gr.Textbox(
        label="Research Question",
        placeholder="What should I verify before applying for a scholarship?",
        lines=2
    )

    urls = gr.Textbox(
        label="Website URLs (one per line)",
        placeholder="https://example.com\nhttps://example.org",
        lines=4
    )

    pdf_files = gr.File(
        label="Upload Multiple PDF Files",
        file_count="multiple",
        file_types=[".pdf"],
        type="filepath"
    )

    button = gr.Button(
        "🔍 Start Investigation",
        variant="primary",
        size="lg"
    )

    output = gr.Markdown(
        label="Investigation Report"
    )

    button.click(
        fn=investigate,
        inputs=[topic, urls, pdf_files],
        outputs=output
    )

    gr.Markdown("""
    ---
    **TruthLens AI | Hackathon 2026**

    Powered by Python, Sentence Transformers and FAISS.
    """)

print("TruthLens AI is ready!")

demo.launch(share=True)
```
